**Обработка пропусков в колонках `eu_sales` и `jp_sales`**

```python
# Выведем первые 5 строк с пропусками в колонке 'eu_sales'
missing_in_eu_sales = df.loc[df['eu_sales'].isna(), ['eu_sales']]
display(missing_in_eu_sales.head(5))
```

```python
# Выведем первые 5 строк с пропусками в колонке 'jp_sales'
missing_in_jp_sales = df.loc[df['jp_sales'].isna(), ['jp_sales']]
display(missing_in_jp_sales.head())
```

```python
# Обработаем пропуски в колонках 'eu_sales' и 'jp_sales'
def pass_processing_1(df, coll_name_1):
    # Вычисляем среднее значение по платформе и году
    group_mean = df.groupby(['platform', 'year_of_release'])[coll_name_1].mean()
    # Среднее по платформе
    platform_mean = df.groupby('platform')[coll_name_1].mean()
    # Среднее по колонке
    coll_mean = df[coll_name_1].mean()

    def staff_row_1(r):
        # Если значение не пропущено, возвращаем его
        if not pd.isna(r[coll_name_1]):
            return r[coll_name_1]

        # Получаем платформу и год для текущей строки
        platform_value = r['platform']
        year_value = r['year_of_release']

        if not pd.isna(platform_value) and not pd.isna(year_value):
            # Ищем среднее значение для данной комбинации платформы и года
            mean_value = group_mean.loc[(platform_value, year_value)]
            if not pd.isna(mean_value):
                return mean_value
        
        if not pd.isna(platform_value):
            mean_value = platform_mean.loc[platform_value]
            if not pd.isna(mean_value):
                return mean_value
            
        return coll_mean
    
    # Применяем функцию к каждой строке
    df[coll_name_1] = df.apply(staff_row_1, axis=1)
    return df

# Заполняем пропуски в колонках с информацией о продахаж
sales_list = ['eu_sales', 'jp_sales']

for coll in sales_list:
    if coll in df.columns:
        df = pass_processing_1(df, coll)
```

```python
# Функция для вывода уникальных значений для каждого столбца ('genre', 'platform', 'rating', 'year_of_release')
def unique_categorical_data(columns_data):
    # Получаем уникальные значения
    print(columns_data.upper())
    unique_val = df[columns_data].unique()
    print(f'Количество уникальных значений: {len(unique_val)}')
    print(f'Уникальные значения: {unique_val}')
    print()


columns_data = ['genre', 'platform', 'rating', 'year_of_release']

for coll in columns_data:
    if coll in df.columns:
        unique_categorical_data(coll)
```

```python
# Функция для поиска неявных дубликатов
def implicit_duplicates(columns_name):

    column_data = df[columns_name]

    original_unique = column_data.unique()
    original_count = len(original_unique)

    normalized= column_data.astype(str).str.lower().str.strip()
    normalized_unique = normalized.unique()
    normalized_count = len(normalized_unique)

    print(columns_name.upper())
    print(f'Оригинальные уникальные значения {original_count}:')
    print(f'{sorted(original_unique)}')

    print(f'Уникальные нормализованные значения {normalized_count}:')
    print(f'{sorted(normalized_unique)}')
    print()
    

    if original_count > normalized_count:
        print(f'Вывод: {original_count - normalized_count} - количество неявных дубликатов')
    else:
        print('Вывод: Неявных дубликатов нет')

    print()
    print()


columns_name = ['genre', 'platform', 'rating']

for coll in columns_name:
    if coll in df.columns:
        implicit_duplicates(coll)
```

## 4. Категоризация данных

```python
# Создаеи копию датасета для работы с данными
df_actual = df_actual.copy()
```

```python
# Создаем функцию для категоризации оценок пользователей
def categorize_user_score(score):

    if score >= 8 and score <= 10:
        return 'Высокая оценка'
    elif score >= 3 and score < 8:
        return 'Средняя оценка'
    elif score >= 0 and score < 3:
        return 'Низкая оценка'

# Создаем новую колонку с категориями
df_actual['user_score_category'] = df_actual['user_score'].apply(categorize_user_score)
```

```python
# Создаем функцию для категоризации оценок критиков
def categorize_critic_score(score):

    if score >= 80 and score <= 100:
        return 'Высокая оценка'
    elif score >= 30 and score < 80:
        return 'Средняя оценка'
    elif score >= 0 and score < 30:
        return 'Низкая оценка'

# Создаем новую колонку с категориями
df_actual['critic_score_category'] = df_actual['critic_score'].apply(categorize_critic_score)
```

```python
# Группировка данных по категориям оценок пользователей
groupby_user_score_category = df_actual.groupby('user_score_category')['user_score'].count()
display(groupby_user_score_category)
```

```python
# Группировка данных по категориям оценок критиков
groupby_critic_score_category = df_actual.groupby('critic_score_category')['critic_score'].count()
display(groupby_critic_score_category)
```
